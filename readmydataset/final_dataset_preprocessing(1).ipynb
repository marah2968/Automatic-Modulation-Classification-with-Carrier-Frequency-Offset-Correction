```python
import pandas as pd
import numpy as np

def generate_one_hot_pattern(modulation_type, num_types=11):
    pattern = [0] * num_types
    modulation_index = int(modulation_type)
    if 0 <= modulation_index < num_types:
        pattern[modulation_index] = 1
    return pattern


modulation_types = df_read['modulation_type'].unique()
num_types = len(modulation_types)
modulation_mapping = {mod_type: idx for idx, mod_type in enumerate(modulation_types)}

df_read['modulation_type_index'] = df_read['modulation_type'].map(modulation_mapping)

print("Modulation type mapping:", modulation_mapping)
print("Unique values in 'modulation_type_index':", df_read['modulation_type_index'].unique())

df_read['one_hot_encoder'] = df_read['modulation_type_index'].apply(lambda x: generate_one_hot_pattern(x, num_types))

print(df_read['one_hot_encoder'])
```

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense,Dropout
import tensorflow as tf
def create_fcn(input_dim):
    inputs = Input(shape=(input_dim,))
    hidden1 = Dense(500, activation='relu')(inputs)
    dropout1 = Dropout(0.2)(hidden1)
    hidden2 = Dense(600, activation='relu')(dropout1)
    dropout2 = Dropout(0.2)(hidden2)
    hidden3= Dense(300, activation='relu')(dropout2)
    dropout3 = Dropout(0.2)(hidden3)
    outputs = Dense(1, activation='linear')(dropout3)
    model = Model(inputs=inputs, outputs=outputs)
    return model
```
